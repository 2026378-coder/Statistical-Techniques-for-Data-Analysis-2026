```python
sample_size=[5, 10, 15, 20, 25, 30]
for size in sample_size:
    means=[]
    for i in range(500):
        sample=data['WorkHoursPerWeek'].sample(n=size, replace=True)
        mean_list_wh=sample.mean()
        means.append(mean_list_wh)
    
    means
    sns.histplot(means ,kde=True)
    dist_mean= sum(means)/len(means)
    plt.title(f'sampling distribution n= {size} and mean_list_wh= {dist_mean}')
    plt.show()
```
