# Measures of central tendency /WorkHoursPerWeek/

```python
mean_wh=data['WorkHoursPerWeek'].mean()
median_wh=data['WorkHoursPerWeek'].median()
mode_wh=data['WorkHoursPerWeek'].mode()
print ('Mean_wh' ,mean_wh)
print ('Median_wh' ,median_wh)
print ('Mode_wh' ,mode_wh)
```

### Plot the distribution /WorkHoursPerWeek/

```python
data_wh= data['WorkHoursPerWeek']

sns.distplot(data_wh, bins=10, hist=True, kde=True, label = 'WorkHoursPerWeek (in millions)')
```

# Computation of measures of dispersion or variability  
# /WorkHoursPerWeek/


```python
min_wh=data['WorkHoursPerWeek'].min()
max_wh=data['WorkHoursPerWeek'].max()
range_wh=data['WorkHoursPerWeek'].max() - data['WorkHoursPerWeek'].min()
var_wh=data['WorkHoursPerWeek'].var()
std_wh=data['WorkHoursPerWeek'].std()
print ('Min_wh' ,min_wh)
print ('Max_wh' ,max_wh)
print ('Range_wh' ,range_wh)
print ('Var_wh' ,var_wh)
print ('Std_wh' ,std_wh)
```

```python
Q1_wh =data['WorkHoursPerWeek'].quantile(0.25)
Q3_wh =data['WorkHoursPerWeek'].quantile(0.75)
IQR_wh = Q3_wh - Q1_wh
lower_bound_wh = Q1_wh - 1.5 * IQR_wh
upper_bound_wh = Q3_wh + 1.5 * IQR_wh
outliers_wh = data[(data['WorkHoursPerWeek'] < lower_bound_wh) | (data['WorkHoursPerWeek'] > upper_bound_wh)]

print ('Q1_wh', Q1_wh )
print ('Q3_wh', Q3_wh )
print ('IQR_wh', IQR_wh )
print ('Lower_bound_wh',lower_bound_wh )
print ('Upper_bound_wh',upper_bound_wh )
print ('Outliers_wh',outliers_wh )
```
